# Chapter 02: Adding tools (notebook)

The same code as `main.py`, one step per cell. Keys: with no key in the environment this notebook replays the chapter's recorded fixtures (`tests/fixtures/replay`), no network. With `OPENAI_API_KEY` or the Azure variables set it runs live. The prompts below are the exact strings the fixtures were recorded with; change one and a replay run reports a missing fixture instead of an answer.

```python
import pathlib
import sys

# Find the repo root (the directory holding tutorials/_shared) and put this chapter on sys.path,
# so the cells below import the chapter's own main.py rather than a copy of it.
HERE = pathlib.Path.cwd().resolve()
ROOT = next(p for p in (HERE, *HERE.parents) if (p / "tutorials" / "_shared").is_dir())
sys.path.insert(0, str(ROOT))
sys.path.insert(0, str(HERE))

import main  # noqa: E402 - the chapter's main.py: its INSTRUCTIONS, tools and helpers are what these cells run
```

## The instructions

They name the tool, so the model knows when to call it.

```python
print(main.INSTRUCTIONS)
```

## The tool

`@tool` turns the function into a name, a description and a JSON schema the model sees. These schemas are part of the replay key.

```python
import json

print(main.get_product_price.name)
print(main.get_product_price.description)
print(json.dumps(main.get_product_price.parameters(), indent=2))
```

## Call the tool directly

No model involved. This is the canned data the model's answer will quote.

```python
print(main.get_product_price.func("SKU-001"))
print(main.get_product_price.func("SKU-999"))
```

## Let the model call it

The model decides to call `get_product_price`, MAF runs it, and the model writes the answer from the result.

```python
agent = main.build_agent()
print(main.DEFAULT_QUESTION)
answer = await main.ask(agent, main.DEFAULT_QUESTION)
print(answer)
assert "79.99" in answer
```
