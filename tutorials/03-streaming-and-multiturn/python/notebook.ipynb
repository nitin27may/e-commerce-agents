## Build the agent

The instructions are the chapter's own.

```python
print(main.INSTRUCTIONS)
agent = main.build_agent()
```

## Two turns on one session

`main.chat()` creates one `AgentSession` and streams each answer chunk by chunk, so the second question can say "it". The questions are the ones the fixtures were recorded with.

```python
questions = [
    "What is Python in one line?",
    "What year was it first released? Answer with the year only.",
]
per_turn = await main.chat(agent, questions)
```

## Inspect the chunks

Each inner list is one turn's streamed chunks. The follow-up only resolves "it" because the session carried turn one.

```python
for question, chunks in zip(questions, per_turn):
    print(f"{len(chunks):>3} chunks  {question}")
second = "".join(per_turn[1])
assert "1991" in second, second
```
